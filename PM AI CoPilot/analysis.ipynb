```python
con.execute("""
SELECT
    CASE
        WHEN maximum_installs < 1000 THEN '<1K'
        WHEN maximum_installs < 10000 THEN '1K-10K'
        WHEN maximum_installs < 100000 THEN '10K-100K'
        WHEN maximum_installs < 1000000 THEN '100K-1M'
        WHEN maximum_installs < 10000000 THEN '1M-10M'
        WHEN maximum_installs < 100000000 THEN '10M-100M'
        ELSE '100M+'
    END AS install_band,

    COUNT(*) AS apps,

    ROUND(
        100.0 * COUNT(*) /
        SUM(COUNT(*)) OVER (),
        2
    ) AS pct_apps
FROM google_play
WHERE maximum_installs IS NOT NULL
GROUP BY install_band
ORDER BY
    CASE install_band
        WHEN '<1K' THEN 1
        WHEN '1K-10K' THEN 2
        WHEN '10K-100K' THEN 3
        WHEN '100K-1M' THEN 4
        WHEN '1M-10M' THEN 5
        WHEN '10M-100M' THEN 6
        WHEN '100M+' THEN 7
    END;
            """).df()
```

```python
con.execute("""
SELECT
    category,

    COUNT(*) AS apps,

    MEDIAN(maximum_installs) AS median_installs,

    AVG(maximum_installs) AS avg_installs,

    MAX(maximum_installs) AS max_installs

FROM google_play

WHERE maximum_installs IS NOT NULL

GROUP BY category

ORDER BY median_installs DESC;
            """).df()
```

```python
con.execute("""
SELECT
    category,

    COUNT(*) AS app_count,

    MEDIAN(maximum_installs) AS median_installs,

    SUM(maximum_installs) AS total_installs,

    ROUND(
        SUM(maximum_installs) * 1.0 /
        COUNT(*),
        0
    ) AS installs_per_app

FROM google_play

WHERE maximum_installs IS NOT NULL

GROUP BY category

ORDER BY installs_per_app DESC;
            """).df()
```

```python
con.execute("""
CREATE OR REPLACE VIEW apps_clean AS

SELECT
    *,
    
    CASE
        WHEN maximum_installs < 1000
            THEN '0-1K'

        WHEN maximum_installs < 10000
            THEN '1K-10K'

        WHEN maximum_installs < 100000
            THEN '10K-100K'

        WHEN maximum_installs < 1000000
            THEN '100K-1M'

        WHEN maximum_installs < 10000000
            THEN '1M-10M'

        WHEN maximum_installs < 100000000
            THEN '10M-100M'

        ELSE '100M+'
    END AS install_band,

    CASE
        WHEN rating > 0
             AND rating_count > 0
        THEN TRUE
        ELSE FALSE
    END AS has_valid_rating

FROM google_play;
            """)
```

```python
con.execute("""
SELECT
    install_band,

    COUNT(*) AS apps,

    COUNT(*) FILTER (
        WHERE has_valid_rating
    ) AS rated_apps,

    ROUND(
        100.0 *
        COUNT(*) FILTER (WHERE has_valid_rating)
        / COUNT(*),
        2
    ) AS pct_rated,

    ROUND(
        AVG(rating) FILTER (WHERE has_valid_rating),
        2
    ) AS avg_rating,

    MEDIAN(rating) FILTER (
        WHERE has_valid_rating
    ) AS median_rating,

    MEDIAN(rating_count) FILTER (
        WHERE has_valid_rating
    ) AS median_rating_count

FROM apps_clean

GROUP BY install_band

ORDER BY
    CASE install_band
        WHEN '0-1K' THEN 1
        WHEN '1K-10K' THEN 2
        WHEN '10K-100K' THEN 3
        WHEN '100K-1M' THEN 4
        WHEN '1M-10M' THEN 5
        WHEN '10M-100M' THEN 6
        WHEN '100M+' THEN 7
    END;
            """).df()
```

```python
# Free vs Paid
con.execute("""
SELECT
    CASE
        WHEN maximum_installs >= 1000000
            THEN '1M+'
        ELSE '<1M'
    END AS adoption_segment,

    free,

    COUNT(*) AS apps,

    ROUND(
        100.0 * COUNT(*)
        / SUM(COUNT(*)) OVER (
            PARTITION BY
                CASE
                    WHEN maximum_installs >= 1000000
                        THEN '1M+'
                    ELSE '<1M'
                END
        ),
        2
    ) AS pct_within_segment,

    MEDIAN(maximum_installs) AS median_installs

FROM google_play

GROUP BY
    adoption_segment,
    free

ORDER BY
    adoption_segment,
    free;
""").df()
```

```python
# Monetization
con.execute("""
SELECT
    CASE
        WHEN maximum_installs >= 1000000
            THEN '1M+'
        ELSE '<1M'
    END AS adoption_segment,

    in_app_purchases,

    COUNT(*) AS apps,

    ROUND(
        100.0 * COUNT(*)
        / SUM(COUNT(*)) OVER (
            PARTITION BY
                CASE
                    WHEN maximum_installs >= 1000000
                        THEN '1M+'
                    ELSE '<1M'
                END
        ),
        2
    ) AS pct_within_segment

FROM google_play

GROUP BY
    adoption_segment,
    in_app_purchases

ORDER BY
    adoption_segment,
    in_app_purchases;
   """ ).df()
```

```python
# Pricing
con.execute("""
SELECT
    CASE
        WHEN maximum_installs >= 1000000
            THEN '1M+'
        ELSE '<1M'
    END AS adoption_segment,

    COUNT(*) AS apps,

    MEDIAN(price) AS median_price,

    AVG(price) AS avg_price,

    MAX(price) AS max_price

FROM google_play

WHERE free = FALSE

GROUP BY adoption_segment;
""").df()
```

```python
# Developer analytics
con.execute("""
SELECT
    developer_id,
    COUNT(*) AS app_count
FROM google_play
WHERE developer_id IS NOT NULL
GROUP BY developer_id
ORDER BY app_count DESC
LIMIT 20;
            """).df()
```

```python
# Category Breakout
con.execute("""
SELECT
    category,

    COUNT(*) AS total_apps,

    COUNT(*) FILTER (
        WHERE maximum_installs >= 1000000
    ) AS breakout_apps,

    ROUND(
        100.0 *
        COUNT(*) FILTER (
            WHERE maximum_installs >= 1000000
        )
        / COUNT(*),
        2
    ) AS breakout_rate,

    MEDIAN(maximum_installs) AS median_installs

FROM google_play

GROUP BY category

HAVING COUNT(*) >= 1000

ORDER BY breakout_rate DESC;
""").df()
```

## Product Opportunity and scale

```python
# recency updates
con.execute("""
SELECT
    last_updated,
    COUNT(*) AS apps
FROM google_play
GROUP BY last_updated
ORDER BY apps DESC
LIMIT 20;
""").df()
```

```python
# Developer scale 
con.execute("""
WITH developer_apps AS (

    SELECT
        developer_id,
        COUNT(*) AS developer_app_count
    FROM google_play
    WHERE developer_id IS NOT NULL
    GROUP BY developer_id

),

apps AS (

    SELECT
        g.*,
        d.developer_app_count
    FROM google_play g
    LEFT JOIN developer_apps d
        ON g.developer_id = d.developer_id
)

SELECT

    CASE
        WHEN developer_app_count = 1 THEN '1 app'
        WHEN developer_app_count BETWEEN 2 AND 5 THEN '2-5 apps'
        WHEN developer_app_count BETWEEN 6 AND 10 THEN '6-10 apps'
        WHEN developer_app_count BETWEEN 11 AND 50 THEN '11-50 apps'
        ELSE '50+ apps'
    END AS developer_segment,

    COUNT(*) AS apps,

    COUNT(*) FILTER (
        WHERE maximum_installs >= 1000000
    ) AS breakout_apps,

    ROUND(
        100.0 *
        COUNT(*) FILTER (
            WHERE maximum_installs >= 1000000
        ) / COUNT(*),
        2
    ) AS breakout_rate

FROM apps

GROUP BY 1

ORDER BY breakout_rate DESC;
""").df()
```

```python
# Category Opportunity Matrix
con.execute("""
SELECT
    category,

    COUNT(*) AS app_count,

    MEDIAN(maximum_installs) AS median_installs,

    QUANTILE_CONT(maximum_installs, 0.75)
        AS p75_installs,

    QUANTILE_CONT(maximum_installs, 0.90)
        AS p90_installs,

    ROUND(
        100.0 *
        COUNT(*) FILTER (
            WHERE maximum_installs >= 100000
        ) / COUNT(*),
        2
    ) AS pct_100k_plus,

    ROUND(
        100.0 *
        COUNT(*) FILTER (
            WHERE maximum_installs >= 1000000
        ) / COUNT(*),
        2
    ) AS breakout_rate,

    MEDIAN(rating) FILTER (
        WHERE rating > 0
          AND rating_count > 0
    ) AS median_rating,

    MEDIAN(rating_count) FILTER (
        WHERE rating > 0
          AND rating_count > 0
    ) AS median_rating_count

FROM google_play

GROUP BY category

HAVING COUNT(*) >= 5000

ORDER BY median_installs DESC;
""").df()
```

### Final Opportunity Score

```python
con.execute("""
CREATE OR REPLACE VIEW category_opportunity AS

WITH category_metrics AS (

    SELECT
        category,

        COUNT(*) AS app_count,

        MEDIAN(maximum_installs) AS median_installs,

        100.0 *
        COUNT(*) FILTER (
            WHERE maximum_installs >= 1000000
        ) / COUNT(*) AS breakout_rate,

        100.0 *
        COUNT(*) FILTER (
            WHERE maximum_installs >= 100000
        ) / COUNT(*) AS pct_100k_plus,

        MEDIAN(rating) FILTER (
            WHERE rating > 0
              AND rating_count > 0
        ) AS median_rating

    FROM google_play

    GROUP BY category

    HAVING COUNT(*) >= 5000
),

ranked AS (

    SELECT
        *,

        PERCENT_RANK() OVER (
            ORDER BY median_installs
        ) AS demand_score,

        PERCENT_RANK() OVER (
            ORDER BY breakout_rate
        ) AS breakout_score,

        PERCENT_RANK() OVER (
            ORDER BY pct_100k_plus
        ) AS scale_score,

        PERCENT_RANK() OVER (
            ORDER BY median_rating
        ) AS quality_score,

        1 - PERCENT_RANK() OVER (
            ORDER BY app_count
        ) AS supply_score

    FROM category_metrics
)

SELECT

    category,

    app_count,

    ROUND(median_installs, 0) AS median_installs,

    ROUND(breakout_rate, 2) AS breakout_rate,

    ROUND(pct_100k_plus, 2) AS pct_100k_plus,

    ROUND(median_rating, 2) AS median_rating,

    ROUND(
          100 * (
              0.30 * demand_score
            + 0.25 * breakout_score
            + 0.20 * scale_score
            + 0.15 * quality_score
            + 0.10 * supply_score
          ),
          2
    ) AS opportunity_score

FROM ranked

ORDER BY opportunity_score DESC;
""").df()
```

### Demand to Supply efficiency

```python
con.execute("""
SELECT

    category,

    COUNT(*) AS app_count,

    MEDIAN(maximum_installs) AS median_installs,

    ROUND(
        MEDIAN(maximum_installs) /
        COUNT(*),
        2
    ) AS demand_supply_efficiency,

    ROUND(
        100.0 *
        COUNT(*) FILTER (
            WHERE maximum_installs >= 1000000
        ) / COUNT(*),
        2
    ) AS breakout_rate

FROM google_play

GROUP BY category

HAVING COUNT(*) >= 5000

ORDER BY demand_supply_efficiency DESC;
""").df()
```

### Simulation Market Distribution

```python
con.execute("""
SELECT
    CASE
        WHEN maximum_installs < 1000 THEN '<1K'
        WHEN maximum_installs < 10000 THEN '1K-10K'
        WHEN maximum_installs < 100000 THEN '10K-100K'
        WHEN maximum_installs < 1000000 THEN '100K-1M'
        WHEN maximum_installs < 10000000 THEN '1M-10M'
        WHEN maximum_installs < 100000000 THEN '10M-100M'
        ELSE '100M+'
    END AS install_band,

    COUNT(*) AS apps,

    ROUND(
        100.0 * COUNT(*) /
        SUM(COUNT(*)) OVER (),
        2
    ) AS pct_apps,

    MEDIAN(rating) AS median_rating,

    MEDIAN(rating_count) AS median_rating_count

FROM google_play

WHERE category = 'Simulation'

GROUP BY install_band

ORDER BY
    CASE install_band
        WHEN '<1K' THEN 1
        WHEN '1K-10K' THEN 2
        WHEN '10K-100K' THEN 3
        WHEN '100K-1M' THEN 4
        WHEN '1M-10M' THEN 5
        WHEN '10M-100M' THEN 6
        WHEN '100M+' THEN 7
    END;
""").df()
```

### Monetization

```python
con.execute("""
SELECT
    in_app_purchases,
    COUNT(*) AS apps,

    ROUND(
        100.0 * COUNT(*) /
        SUM(COUNT(*)) OVER (),
        2
    ) AS pct_apps,

    MEDIAN(maximum_installs) AS median_installs,

    MEDIAN(rating_count) AS median_rating_count

FROM google_play

WHERE category = 'Simulation'

GROUP BY in_app_purchases
ORDER BY median_installs DESC;
""").df()
```

### Simulation Breakout vs Non Breakout

```python
con.execute("""
WITH simulation_apps AS (

    SELECT
        *,
        CASE
            WHEN maximum_installs >= 1000000
            THEN 'Breakout'
            ELSE 'Non-Breakout'
        END AS adoption_segment

    FROM google_play

    WHERE category = 'Simulation'
)

SELECT

    adoption_segment,

    COUNT(*) AS apps,

    ROUND(AVG(rating), 2) AS avg_rating,

    MEDIAN(rating_count) AS median_rating_count,

    MEDIAN(maximum_installs) AS median_installs,

    ROUND(
        100.0 * COUNT(*) FILTER (
            WHERE in_app_purchases = TRUE
        ) / COUNT(*),
        2
    ) AS pct_iap,

    ROUND(
        100.0 * COUNT(*) FILTER (
            WHERE free = TRUE
        ) / COUNT(*),
        2
    ) AS pct_free

FROM simulation_apps

GROUP BY adoption_segment;
""").df()
```
