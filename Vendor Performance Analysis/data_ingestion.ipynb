```python
import pandas as pd
import os
import duckdb #parallel processing (batches), No chunking required, fast csv ingestion as compared to SQLite 
import time
```

```python
con = duckdb.connect('my_db.duckdb')
```

```python
def raw_data_to_db():
    for file in os.listdir('data'):
        if '.csv' in file:
            start = time.time()
            print('\n----------------------------------------------------\n')
            print(f'Starting: {file}')
            
            con.execute(f"""
                CREATE TABLE {file[:-4]} AS 
                SELECT * FROM read_csv_auto('data/{file}')  
            """)
            # read_csv_auto Reads in parallel, Uses vectorized execution, Much faster than pandas

            print(f'Data Ingested to Table: {file[:-4]}')
            end = time.time()
            time_taken = (end - start)/60
            print(f'Total Ingestion time for {file[:-4]} : {time_taken} minutes')
```

```python
raw_data_to_db()
```

```python
con.close()
```
