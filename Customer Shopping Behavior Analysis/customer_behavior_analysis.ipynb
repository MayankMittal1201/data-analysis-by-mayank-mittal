### Feature Engineering

```python
# labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
# df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)

bins = [0, 25, 40, 60, 100]
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']

df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

### SQL ANALYSIS

```python
import duckdb
```

```python
con = duckdb.connect()
```

```python
con.register('user_shopping_data', df)
```

```python
con.execute("""
select * 
from user_shopping_data 
limit 5 
            """).fetch_df()
```

```python
# total revenue generated by Male vs Female customers

con.execute("""
select gender, SUM(purchase_amount) as revenue
from user_shopping_data
group by gender
            """).fetch_df()
```

```python
# customers used a discount but still spent more than the average purchase amount
con.execute("""
select customer_id, purchase_amount 
from user_shopping_data 
where discount_applied = 'Yes' and purchase_amount >= (select AVG(purchase_amount) from user_shopping_data)
""").fetch_df()
```

```python
# top 5 products with the highest average review rating
con.execute("""
select item_purchased, round(avg(cast(review_rating as float)),2) as "Average Product Rating"
from user_shopping_data
group by item_purchased
order by avg(review_rating) desc
limit 5
""").fetch_df()
```

```python
# average Purchase Amounts between Standard and Express Shipping
con.execute("""
select shipping_type, round(avg(purchase_amount),2) avg_pur_amt
from user_shopping_data
where shipping_type in ('Standard','Express')
group by shipping_type
""").fetch_df()
```

```python
# average spend and total revenue - subscribers vs non-subscribers
con.execute("""
select subscription_status,
    count(customer_id) as total_customers,
    round(avg(purchase_amount),2) as avg_spend,
    round(sum(purchase_amount),2) as total_revenue
from user_shopping_data
group by subscription_status
order by total_revenue desc
""").fetch_df()
```

```python
# Top 5 products with the highest percentage of purchases with discounts applied
con.execute("""
select item_purchased,
       round(100.0 * sum(case when discount_applied = 'Yes' then 1 else 0 end)/count(*),2) as discount_rate
from user_shopping_data
group by item_purchased
order by discount_rate desc
limit 5
""").fetch_df()
```

```python
# New, Returning, and Loyal
con.execute("""
with customer_type as (
select customer_id, previous_purchases,
case 
    when previous_purchases = 1 then 'New'
    when previous_purchases between 2 and 10 then 'Returning'
    else 'Loyal'
    end as customer_segment 
from user_shopping_data)

select customer_segment,count(*) as no_of_customers
from customer_type 
group by customer_segment
""").fetch_df()
```

```python
# Top 3 most purchased products within each category
con.execute("""
with item_counts as (
    select category,
           item_purchased,
           count(customer_id) as total_orders,
           row_number() over (partition by category order by count(customer_id) desc) as item_rank
    from user_shopping_data
    group by category, item_purchased
)
select item_rank,category, item_purchased, total_orders
from item_counts
where item_rank <=3
""").fetch_df()
```

```python
# repeat buyers (more than 5 previous purchases) also likely to subscribe
con.execute("""
select subscription_status, count(customer_id) as repeat_buyers
from user_shopping_data
where previous_purchases > 5
group by subscription_status
""").fetch_df()
```

```python
# revenue contribution of each age group
con.execute("""
select age_group, 
    count(customer_id) as total_customers, 
    round(avg(purchase_amount), 2) as avg_pur_amt, 
    sum(purchase_amount) as total_revenue
from user_shopping_data
group by age_group
order by total_revenue desc
""").fetch_df()
```
